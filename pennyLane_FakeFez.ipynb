```python
import pennylane as qml
from qiskit_ibm_runtime.fake_provider import FakeFez

backend = FakeFez()
dev = qml.device("qiskit.remote", wires=2, backend=backend)   # 2, not 127
```

```python
@qml.qnode(dev)
def circuit():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.probs(wires=[0, 1])
```

```python
dev = qml.device("qiskit.remote", wires=2, backend=backend, shots=500)

@qml.qnode(dev)
def circuit():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.probs(wires=[0, 1])

result = circuit()
print(result)
```
