```python
from qiskit import QuantumCircuit, transpile
# Import the specific fake backend
from qiskit_ibm_runtime.fake_provider import FakeFez
```

```python
# 1. Instantiate the fake backend
backend = FakeFez()
```

```python
# 2. Build your abstract circuit (e.g., a Bell State)
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
```

```python
# 3. Transpile the circuit (Crucial Step!)
# This rewrites your circuit using Fez's native basis gates and
# routes your logical qubits to specific physical qubits on the 156-qubit chip.
transpiled_qc = transpile(qc, backend=backend)
```

```python
# 4. Execute the noisy simulation
# Because you have qiskit-aer installed, this automatically injects
# the Fez noise model into the simulation.
job = backend.run(transpiled_qc, shots=500)
result = job.result()
```

```python
# 5. View the realistic results
counts = result.get_counts()
print(counts)
```
