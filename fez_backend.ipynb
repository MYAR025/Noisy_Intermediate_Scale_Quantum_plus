```python
from qiskit import QuantumCircuit, transpile
# Import the specific fake backend
from qiskit_ibm_runtime.fake_provider import FakeFez
```

```python
service = QiskitRuntimeService()
backend = service.backend("ibm_fez")
```

```python
# 2. Build your abstract circuit (e.g., a Bell State)
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()
```

```python
# 3. Transpile the circuit (Crucial Step!)
# This rewrites your circuit using Fez's native basis gates and
# routes your logical qubits to specific physical qubits on the 156-qubit chip.
transpiled_qc = transpile(qc, backend=backend)
```

```python
from qiskit_ibm_runtime import SamplerV2 as Sampler

sampler = Sampler(mode=backend)
job = sampler.run([transpiled_qc], shots=500)
result = job.result()

counts = result[0].data.meas.get_counts()
print(counts)
```
